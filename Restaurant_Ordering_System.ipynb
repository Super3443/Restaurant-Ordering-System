```python
order_list = [] # For only the current order
order_history = [] # Stores all the orders made
points = 0 # Stores points for current order
total_points = 0 # Stores points from every order
has_membership = None # Initialise membership status as None
    
def food_lists():
    # Main menu items and prices
    main_items = ["CheeseBurger", "Fish & Chips", "Hamburg Steak", "Fried Rice"] # List of Main food
    main_prices = [6.50, 7.00, 9.50, 10.50] # List of price for Main food

    # Side menu items and prices
    side_items = ["Fries", "Ice Cream", "Strawberry Cheesecake"]  # List of Main food
    side_prices = [3.00, 1.50, 4.50] # List of price for Side food
    
    return main_items, main_prices, side_items, side_prices

# Returns the type of food and the quantity
def determine_food(main_food, side_food):
    main_items, main_prices, side_items, side_prices = food_lists()
    
    main_dish = None # To be updated with the main dish
    side_dish = None # To be updated with the side dish
    
    if main_food is not None and 0 <= main_food < len(main_items): # When main dish was selected and checks if the input was valid
        main_dish = main_items[main_food]
        
    if side_food is not None and 0 <= side_food < len(side_items): # When side dish was selected and checks if the input was valid
        side_dish = side_items[side_food]
        
    return main_dish, side_dish

# Displays each order summary
def print_orders(item_details):
    # Display items
    grand_total = 0 # Initialise the total for the entire order
    order_counter = 1 # Initialise the counter
    
    # Print out the summary of the order
    for item in item_details:
        dish_type, name, quantity, price, total = item
        print(f"Item {order_counter}:")
        print(f"{dish_type} Dish: {name}")
        print(f"Quantity: {quantity}")
        print(f"Price Each: ${price:.2f}")
        print(f"Item Total: ${total:.2f}")
        print()
        grand_total += total
        order_counter += 1
            
    print(f"Order Total: ${grand_total:.2f}\n")


# Calculates the total of each dish and the total of the current order
def price_calculator():
    items = None
    
    if items is None: # When items is empty
        items = order_list
        
    main_items, main_prices, side_items, side_prices = food_lists()
    # Finding the price
    grand_total = 0 # Initialise the total of all the dishes
    item_details = []  # Will store (price, total) for each item
        
    # Displays all orders
    for order in order_list:
        # Separate the order details
        dish_type, name, quantity = order
        price = 0 # Initialise the price
        
        # For Main Dishes
        if dish_type == "Main":
            for num in range(len(main_items)): # i will iterate from 1,2,3,4,5
                if main_items[num] == name: # Checks each iteration of i if it matches the name of the dish
                    price = main_prices[num] # Get the corresponding price
                    
        # For Side Dishes
        if dish_type == "Side":
            for num in range(len(side_items)): # i will iterate from 1,2,3,4,5
                if side_items[num] == name: # Checks each iteration of i if it matches the name of the dish
                    price = side_prices[num] # Get the corresponding price
        
        total = price * quantity # Calculates the total for the specific dish
        grand_total += total # Calculates the total for all the dishes combined
        item_details.append((dish_type, name, quantity, price, total)) # Append all the info of the dish into item_details

    return item_details, grand_total

def points_calculator(grand_total, has_membership):
    # Check whether use has membership
    if has_membership == True: # When user has membership
        item_details, grand_total = price_calculator()
        points = grand_total * 0.5 # Convert $1 to 0.1 points each
        return round(points)
        
    else: # When the user does not have a membership
        print("No Points given. Please Apply For Membership First")
        return 0
    
# Displays full menu for the user
def view_menu():
    main_items, main_prices, side_items, side_prices = food_lists()
    # To Display the Main menu
    # To read the Main_prices list
    print("=== This Is The Main Menu ===")
    counter = 0
    for i in main_items:
        counter += 1
        if counter >= 1:
            print(f"{counter}. {i} ${main_prices[counter - 1]}")
    print()

    while True:
        side_menu_choice = input("Would You Like To See The Sides? (Input y or n): ").lower() # Prompt user for side menu
        print()
        
        if side_menu_choice == "y": # when side menu is chosen
            print("=== This Is The Side Menu ===")
            # To Display the side menu
            # To read the side_prices list
            counter = 0
            for i in side_items:
                counter += 1
                if counter >= 1:
                    print(f"{counter}. {i} ${side_prices[counter - 1]}")
            print()
            break
            
        elif side_menu_choice == "n": # when side menu is not chosen
            break
                
        else: # When invalid input
            print("Invalid Choice. Please Input y or n ")
            print()

# Adds the current order
def add_order(order_list):
    main_items, main_prices, side_items, side_prices = food_lists() # Get detailed pricing breakdown and pre-discount total
    
    while True:
        # Main Dish Selection
        main_choice = int(input("Please Input Your Main Dish of Choice (e.g 1) or 0 If No Main Dishes: "))
        
        if main_choice == 0:
            main_food = None  # If no Main dish is selected
            print()
            
        elif 1 <= main_choice <= len(main_items): # When main dish is selected
            main_food = main_choice - 1  # To ensure index is read properly since it starts at 0
            main_amount = int(input("Please Input The Quantity For The Main Dish (e.g 1): "))
            print()

            if main_amount <= 0: # When quantity is 0 or lower
                print("Invalid Quantity. Please Input A Positive Number\n")
                continue
            
        else: # When input is invalid
            print("Invalid input. Please Enter A Valid Number (e.g 1) or 0 If No Main Dishes\n")
            continue
              
        # Side Dish Selection
        side_choice = int(input("Please Input Your Side Dish of Choice (e.g 1) or 0 if No Side Dishes: "))
        
        if side_choice == 0:
            side_food = None  # If no Side dish is selected
            print()
                
        elif 1 <= side_choice <= len(side_items): # When side dish is selected
            side_food = side_choice - 1  # To ensure index is read properly since it starts at 0
            side_amount = int(input("Please Input The Quantity For The Side Dish (e.g 1): "))
            print()

            if side_amount <= 0: # When input is invalid
                print("Invalid Quantity. Please Input A Positive Number\n")
                continue

        else: # When input is invalid
            print("Invalid input. Please Enter A Valid Number (e.g 1) or 0 If No Side Dishes\n")
            print()
            continue
                
        if main_food == None and side_food == None: # When no dishes were selected
            print("No Dishes Chosen")
            print()
            return
            
        else:
            main_dish, side_dish = determine_food(main_food, side_food) # Get the main and side dish chosen earlier
            
            print("Selected Items:")
            # Stores the current order, including quantity and type of dish, together into a list
            if main_dish:
                order_list.append(("Main", main_dish, main_amount))  # append the dish_type, main_dish and amount
                print("Main Dish:", main_dish)
                print("Quantity:", main_amount)
                print()
                
            if side_dish:
                order_list.append(("Side", side_dish, side_amount))  # append the dish_type, side_dish and amount
                print("Side Dish:", side_dish)
                print("Quantity:", side_amount)
                print()
                
            return

# Allows update or removal of an item in the current order
def modify_order_list():
    if not order_list:
        print("Order List Is Empty; Unable To Make Any Modifications")
        print()
        return # Prevents user from modifying any order
            
    # Show the user's Order List for easier navigation
    full_list_display()
    main_items, main_prices, side_items, side_prices = food_lists() # Get the lists of food and prices
        
    print("1. Update Item")
    print("2. Remove Item Completely")
    print()

    while True:
        # Prompt whether the user wants to update or remove an item from the order list
        modified_choice = int(input("Do You Want To Update or Remove An Item From The Order List? (Input Respective Number Of The Choice e.g 1) ")) - 1
        print()
        
        # Checks if input is valid
        if modified_choice not in (0, 1): # When input isn't Valid
            print("Invalid Input. Please Enter Either 1 or 2")
            print()
            
        # When input is Valid
        # Prompt the user to decide what item to either update or remove from the order list
        modified_item_name = int(input("Which Item Do You Want To Modify? (Only Input The Item Number In The Order) ")) - 1
        print()
    
        if not 0 <= modified_item_name < int(len(order_list)): # When input is invalid
            print("Invalid Input. Please Enter A Valid Number (e.g 1)")
            print()
                
        else:
            # Get current item details
            dish_type, name, current_quantity = order_list[modified_item_name]
                
            # Checks which choice was made
            if modified_choice == 0: # Update the Item with a new quantity
                    
                # Update quantity
                print("Current quantity:", current_quantity)
                    
                new_quantity = int(input("Enter new quantity (e.g 2): "))
                if new_quantity > 0:
                    order_list[modified_item_name] = (dish_type, name, new_quantity) # Updates the new quantity for the item selected
                    print(f"Updated Quantity From {current_quantity} to {new_quantity}")
                    print()
                    
                else:  # When input is invalid
                    print("Invalid Input. Please Enter A Valid Number (e.g 2)")
                    print()
        
            else: # Remove the item completely
                confirmation = input("Are You Sure You Want To Remove This Item? (Input y or n): ").lower() # Ask to make sure no item is removed accidentaly
                print()
                    
                if confirmation == "y": # When user confirms to remove the item completely
                    removed_item = order_list[modified_item_name]  # Get the item name to display
                    order_list.remove(removed_item)  # Remove by value
                    print(f"Successfully Removed {removed_item[1]} From The Order !") # Display to User the item removed
                    print()
                    return # Prevents user from starting the option again
                    
                elif confirmation == "n": # When user doesn't want to remove item
                    print("Item Not Removed.")
                    print()
                    return # Prevents user from starting the option again
                    
                else:  # When input is invalid
                    print("Invalid Input. Please Enter Either y or n")
                    print()

# Displays the current order
def full_list_display():
    print("=== CURRENT ORDER ===")

    if not order_list:  # Check if order list is empty
        print("No Orders Yet.")
        print()
        
    else:
        item_details, grand_total = price_calculator() # Get the details of each dish and the total of the order
        print_orders(item_details) # print out the order

# For the process of checking out the current order
def checkout(order_history, total_points):
    if not order_list:  # Check if order is empty first
        print("Your Order is Empty. Please Add Items First.\n")
        return order_history, total_points
    
    full_list_display() # Display order summary
    item_details, grand_total = price_calculator() # Get the details of the dish and total of the entire order
    points_earned = points_calculator(grand_total, has_membership) # Get calculated points
    final_total = grand_total # Initialise another total for the entire order. To use for calculations
    points_used = 0 # Initiate the points used
    
    # Points redemption
    if has_membership == True and total_points > 0: # When user has membership and points above 0
        print(f"You have {total_points} points available (${total_points*0.25:.2f} discount)")
        use_points = input("Would you like to use your points? (y/n): ").lower()
        if use_points == 'y':
            max_points = min(total_points, grand_total/0.25) # Find the maximum point that does not exceed the total which can be used 
            points_used = int(input(f"How many points to use? (1-{int(max_points)}): ")) # Ask how many points to use
            if 0 < points_used <= max_points:
                discount = points_used * 0.25
                final_total = grand_total - discount
                total_points -= points_used  # Deduct used points
                print(f"Applied {points_used} points (${discount:.2f} discount)")
                print(f"New total: ${final_total:.2f}")
            else:  # When input is invalid
                print("Invalid points amount. No discount applied.")
        else:  # When input is invalid
            print("Invalid Input. Please Enter Either y or n")
    
    # Confirmation
    confirmation = input("Confirm checkout? (y/n): ").lower()
    if confirmation == 'y':
        new_history_entry = []
        counter = 0
        while counter < len(order_list):
            new_history_entry.append(order_list[counter])
            counter += 1
        order_history.append(new_history_entry)
        
        # Update points (add earned points)
        if has_membership:
            total_points += points_earned
            print(f"Points earned: {points_earned}")
            if points_used > 0:
                print(f"Points used: {points_used}")
            print(f"Total points remaining: {total_points}\n")
        
        order_list.clear()
        payment_option()
        return order_history, total_points
    else:
        print("Checkout Cancelled.\n")
        return order_history, total_points

def payment_option():
    print("1. Google Pay")
    print("2. Apple Pay")
    print("3. Card\n")

    # Prompt user for choice of payment method
    payment_choice = int(input("Please Select Payment Option (e.g 1 for Apple Pay): "))
    print()

    if 1 <= payment_choice <= 3: # Valid option is chosen for payment
        print("Order Completed! Thank You!\n")
        return order_history, total_points

    else:  # When input is invalid
        print("Invalid Input. Please Input A Valid Choice (e.g 1 for Apple Pay)")
            
# Displays all the orders that were made
def view_order_history(order_history):
    print("=== ORDER HISTORY ===")
    
    # Check if empty
    if order_history == []: # When empty
        print("No Order History.\n")
        return

    # Calculation of price without function due to order_list being empty
    else: # When not empty
        order_counter = 1 # Initiatlise the counter
        # Print the specific order
        for order in order_history:
            print(f"\nOrder {order_counter}:")
            print("-" * 20)  # Separator line
            
            # Calculate prices for this historical order
            item_details = []
            main_items, main_prices, side_items, side_prices = food_lists()

            # Iterate through the order to find the total of the entire specific order and each dish
            for item in order:
                dish_type, name, quantity = item
                price = 0
                
                if dish_type == "Main": # When main dish
                    counter = 0 # Initiatlise the counter
                    while counter < len(main_items):
                        if main_items[counter] == name:
                            price = main_prices[counter]
                            break
                        counter += 1
                        
                if dish_type == "Side": # when side dish
                    counter = 0 # Initiatlise the counter
                    while counter < len(side_items):
                        if side_items[counter] == name:
                            price = side_prices[counter]
                            break
                        counter += 1
                
                total = price * quantity # Calculates the total for the specific dish
                item_details.append((dish_type, name, quantity, price, total)) # Append details into a list
            
            # Display items
            item_number = 1 # Initiatlise the counter
            grand_total = 0 # Initiatlise the grand_total

            # Print out each item for the specific order
            for item in item_details:
                dish_type, name, quantity, price, total = item
                print(f"Item {item_number}:")
                print(f"{dish_type} Dish: {name}")
                print(f"Quantity: {quantity}")
                print(f"Price Each: ${price:.2f}")
                print(f"Item Total: ${total:.2f}\n")
                grand_total += total
                item_number += 1
                
            print(f"Order Total: ${grand_total:.2f}\n")
            
            order_counter += 1 # Move to next order
            
# To apply or modify membership status
def view_membership(has_membership):
    while True:
        # Checks if user's membership status
        if has_membership == None: # When user doesn't have membership
            apply_membership = input("Would You Like To Apply For Membership? (Input Either y or n): ").lower()
    
            if apply_membership == "y":
                has_membership = True
                print("Membership Successfully Applied! You Are Now A Member\n")
                return has_membership
        
            elif apply_membership == "n":
                print("Membership Not Applied!\n")
                return has_membership
        
            else:  # When input is invalid
                print("Invalid Input. Please Input Either y or n\n")
    
        else: # When user already has membership
            print("You Are Already A Member\n")
            print("1. View Points")
            print("2. Remove Membership")
            print("3. Go Back\n")
                  
            modify_membership = int(input("Input Your Choice (e.g 1): ")) # Ask which choice to perform

            # Checks which choice was selected
            if modify_membership == 1: # When view points choice was selected
                view_points()
                has_membership = True # Set membership status to true
                return has_membership # Return the membership status

            elif modify_membership == 2: # When remove membership choice was selected
                remove_membership_choice = input("Are You Sure You Want To Remove Membership? (Input Either y or n): ")
                print()

                # Check choice for removing membership
                if remove_membership_choice == "y": # Remove membership is selceted
                    has_membership = None # Set membership status to false
                    print("Membership Successfully Removed! You Are No longer A Member\n")
                    return has_membership # Return the membership status
         
                elif remove_membership_choice == "n": # Remove membership is not selceted
                    print("Membership Not Removed! You Are Still A Member\n")
                    return has_membership # Return the membership status
        
                else: # When input was invalid
                    print("Invalid Input. Please Input Either y or n\n")

            elif modify_membership == 3: # When go back choice was selected
                has_membership = True
                return has_membership

            else: # When input was invalid
                print("Invalid Input. Please Input A Valid Choice (e.g 1)\n")

# View the points
def view_points():
    # Checks if user has membership
    if has_membership == True: 
        print("(Membership Active - Earning Points)")
        print(f"Total Points: {total_points}\n")
    else:
        print("(No Membership - Not Earning Points)\n")

# Main Programme
while True:
    # Display the choices for the user to select
    print("=== Welcome To Python Food Cafè ===")
    print("")
    print("What Would You Like To Do?")
    print("1. View Menu")
    print("2. Add Item Into Order List")
    print("3. Modify Item In The Order List")
    print("4. Display Current Order List")
    print("5. Checkout")
    print("6. Display Order History")
    print("7. Membership")
    print("0. Exit")
    print()
    
    choice = int(input("Please Input Your Choice (e.g 1): ")) # Prompt user to input their choice
    print()
    
    # Show the user what they chose
    # View Menu
    if choice == 1:
        view_menu()
        
    # Add an item to the order list
    elif choice == 2:
        add_order(order_list)

    # To Update Items in the Order List
    elif choice == 3:
        modify_order_list()
        
    # To Display Current Order List
    elif choice == 4:
        full_list_display()

    # To Checkout
    elif choice == 5:
        order_history, total_points = checkout(order_history, total_points)

    # To Display Order History
    elif choice == 6:
        view_order_history(order_history)

    # To Apply For Membership
    elif choice == 7:
        has_membership = view_membership(has_membership)
        
    # To Display when exiting the program
    elif choice == 0:
        print("Thank You For Ordering From Python Food Cafè. Goodbye!")
        break
        
    # To Display when input is invalid
    else:
        print("Invalid Choice. Please Input A Valid Choice (e.g 1)")
        print()
```
